#Tuple
##ordered collecction of items ()
##One main difference with is tuple is immutable
##var_name = list[[stu1:()],[stu2:()]...]

```python
# Create a tuple
student = (101 , "Sanjiv" , "CSE" , 8.9)
print(student)

#tuples using index
print("Student Information in first location:" ,student[0])
print(student[1])

#negative index
print("Example for negative index:" , student[-2])


```

```python
# TO find the length of the tuple
subjects = ("DSA" , "DSV" , "OS" , "Python" , "Machine Learning")
print(len(subjects))
```

```python
#slicing
subjects = ("DSv" , "DSA" , "Statistics" , "SQL" , "Machine Learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])

```

```python
# tuple immutable
students_info = ("Chirag" , 21 , "CSE" , 8.7)
students_info[3] = 9.1
print(students_info)
```

```python
# Nested Tuple
students = (101 , "Rahul" , (89 , 78 , 66))
print (students)
print(students[2])
print(students[2][1])
```

```python
# count function
marks = ( 75 , 55 , 55 , 88 , 99 , 88  , 20 )
print ( "Time of occurance: " , marks.count(55))
```

```python
# find the element present in tuple or not
subjects = ("Python",
            "DSV" ,
            "SQL" ,
            "Machine Learning" ,
            "Statistics")
print("Python" in subjects)
print("JAVA" in subjects)
```

# SET
## unordered collection of items
## mutable
## allows the different data types
## id doesn't support index
## it gives output with uniques (NO duplication)


```python
# Symbol for set {}
# Create a set with similar items
marks = {75 , 68 , 89 , 55 , 78 , 89 , 75 , 66 , 55 , 99}
print(marks)
```

```python
# create a emply set
a = {}
a = set()
# check its type
print(type(a))
```

```python
# set doesn't support the index
students = {"Amit" , "Riya" , "Priya"}
print(students[1])
```

```python
# add an item in the set
students =  { "Amit" , "Kiran" , "Sneha"}
students.add("Priya")
print(students)
students.update(["Ram" , "Chitra" , "Vani" , "Devi"])
print(students)
```

```python
# remove an item from the set
students = {"Amit" , "Riya" , "Priya"}
students.remove("Priya")
print(students)

students.discard("Amit")
print(students)
```

```python
students = {"Amit" , "Riya" , "Priya"}
students.remove("Karan")
print(students)
```

```python
# create a set and clear list of items
students_info = {
    "Amit" ,
    "Karan" ,
    "Riya"
}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks = {99.88 , 77 , 66}
del marks
print(marks)
```

```python
# set operations
python_students={"Amit" , "Neha" , "Sanjiv" , "Manish"}
java_students = {"Amit" , "karan" , "Dhilip" , "Manish" , "Neha"}

# for union operation by using set
result = python_students.union(java_students)
print("Union Operation:" , result)

# for intersection operation by using set
result_intersection = python_students.intersection(java_students)
print("Intersection Operation:" , result_intersection)

# to find difference with two set
result_diff = python_students.difference(java_students)
print(result_diff)
```

```python
# convert list to set
stud_id = [101,102,103,102,103,104,105]
unique_stud_id = set(stud_id)
print(unique_stud_id)
uniq_roll_no = list(unique_stud_id)
print(uniq_roll_no)
```
